```python
import numpy as np
```

```python
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr
```

```python
Q1 = np.percentile(arr,25)
Q3 = np.percentile(arr,75)
```

```python
print(Q1)
print(Q3)
```

```python
IQR = Q3 -Q1
print(IQR)
```

```python
UF = Q3 +(1.5*IQR)
LF = Q1 -(1.5*IQR)

print(UF)
print(LF)
```

```python
print(arr)
```

```python
L=[]
for i in arr:
    if i>=LF and i<=UF:
        L.append(i)
arr2 = np.array(L)

print(arr2)
```

```python
import seaborn as sns 

sns.boxplot(x = arr2)
```
